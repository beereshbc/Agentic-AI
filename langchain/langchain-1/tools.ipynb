### Tools


```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("openai/gpt-oss-120b", model_provider="groq")

res = model.invoke("Tell me about karnataka")
print(res)

```

How to intigrate the tool


```python
from langchain.tools import tool

def getWeather(location:str)-> str:
    """This will provide weather of specific location"""
    return f"The weather in the {location} is sunny with little bit rain"


model_with_tool = model.bind_tools([getWeather])


response = model_with_tool.invoke("What is the weather in Bengaluru")
print(response)

for tool_call in response.tool_calls:
    print(f"Tools : {tool_call["name"]}")
    print(f"Args : {tool_call['args']}")
```

```python
messages = [{"role":"user", "content": "What is the weather in Bengaluru"}]

ai_msg = model_with_tool.invoke(messages)
messages.append(ai_msg)

for tool_call in ai_msg.tool_calls:
    tool_result = getWeather(tool_call)
    messages.append(tool_result)

final_response = model_with_tool.invoke(messages)
print(final_response.text)
```

# Messages


```python
from langchain.messages import SystemMessage, AIMessage, HumanMessage

messages = [
    SystemMessage("Karnataka is Beautiful state"),
    HumanMessage("Tell me the slogan of one state many worlds in karnataka")
] 

res = model.invoke(messages)
print(res.content)
```

```python
sys_msg = SystemMessage("You are helpful coding assistent")

messages = [
    sys_msg,
    HumanMessage("How to write middleware in MERN")
]

res = model.invoke(messages)
print(res)
```

```python
system_msg = SystemMessage("Your Sr. Python Developer with full knowladge of backend develpoment and very good and strong tuter")

messages = [
    system_msg,
    HumanMessage("How to write REST api")
]

res = model.invoke(messages)
print(res)
```

```python
msg = HumanMessage(
    content="guess a number between 1 to 20",
    name="Beeresh",
    id="1"
)

res = model.invoke([msg])
print(res)
```

```python
from langchain.messages import ToolMessage, AIMessage, HumanMessage

ai_message = AIMessage(
    content=[],
    tool_calls=[
        {
            "name":"getWeather",
            "args":{"location":"Davanagere"},
            "id":"_id2"
        }
    ]
)

weather_result = "Sunny, 72' F"
tool_message = ToolMessage(
    content=weather_result,
    tool_call_id = "call_123"
)

messages = [
    HumanMessage("Whats the weather in davanagere"),
    ai_message,
    tool_message,
]

res = model.invoke(messages)
print(res)
```
