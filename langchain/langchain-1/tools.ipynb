### Tools


```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")

model = init_chat_model("openai/gpt-oss-120b", model_provider="groq")

res = model.invoke("Tell me about karnataka")
print(res)

```

How to intigrate the tool


```python
from langchain.tools import tool

def getWeather(location:str)-> str:
    """This will provide weather of specific location"""
    return f"The weather in the {location} is sunny"


model_with_tool = model.bind_tools([getWeather])


response = model_with_tool.invoke("What is the weather in Bengaluru")
print(response)

for tool_call in response.tool_calls:
    print(f"Tools : {tool_call["name"]}")
    print(f"Args : {tool_call['args']}")
```

```python
messages = [{"role":"user", "content": "What is the weather in Bengaluru"}]

ai_msg = model_with_tool.invoke(messages)
messages.append(ai_msg)

for tool_call in ai_msg.tool_calls:
    tool_result = getWeather(tool_call)
    messages.append(tool_result)

final_response = model_with_tool.invoke(messages)
print(final_response.text)
```
