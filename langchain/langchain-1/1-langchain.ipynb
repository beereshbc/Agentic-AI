### Langchain Version V1

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

```python
from langchain.agents import create_agent

def getWeather(city=str) -> str:
    """Get the Weather for city"""
    return "the weather in {city} is sunny"

agent = create_agent(
    model="gpt-6-astra",
    tools=[getWeather],
    system_prompt="A you are a helpfull Assistant",
    debug=True
)

agent
```

```python
agent.invoke({'messages': [{"role":"user", "content":"Whats teh weather in davanagere"}]})
```
